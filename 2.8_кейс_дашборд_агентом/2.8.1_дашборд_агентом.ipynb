Результат — постоянный объект со ссылкой.

## Описываем дашборд словами

Так же, как поставили бы задачу аналитику: какие показатели нужны и в каком виде.

## Приёмка

Откройте ссылку выше и проверьте пять вещей — на это уходит минута, и она
окупается на первом же обсуждении с командой:

1. **Каждая карточка показывает данные, а не сообщение об ошибке.** Пустая или
   красная карточка означает, что запрос не выполнился.
2. **У графика подписаны оси.** Если система не поняла, что по горизонтали, а что
   по вертикали, карточка спросит об этом прямо на дашборде.
3. **Числа совпадают с тем, что вы считали в тетради.** Накопленный NPV портфеля
   на последнем месяце, число инициатив, суммы по направлениям.
4. **Названия понятны без пояснений.** Дашборд смотрят те, кто не присутствовал
   при его сборке.
5. **Относительный месяц не спутан с абсолютным.** В паспорте есть и месяц старта,
   и число месяцев до окупаемости. Столбец «месяц окупаемости» со значением 14
   может означать и четырнадцатый месяц портфеля, и четырнадцатый месяц от старта
   самой инициативы — это разные вещи, и на этом чаще всего расходятся числа
   в отчётах. Посмотрите запрос карточки и убедитесь, что имелось в виду именно то,
   что вы просили.

Что-то не так — правится прямо в интерфейсе: карточка открывается, запрос виден
целиком, тип отображения меняется в два клика. Агент не заменяет вас в этом месте,
он лишь экономит время на первой сборке.
